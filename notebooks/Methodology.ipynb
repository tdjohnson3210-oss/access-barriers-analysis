### Project Description
This project examines how transportation barriers and other structural access constraints influence civil filing access across U.S. communities. By integrating geospatial data, filing patterns, and community‑level indicators, the study identifies where limited transportation, long travel times, digital access gaps, and rural–metro differences shape or suppress filing behavior. The goal is to reveal how these barriers contribute to inequities in civil justice accessibility.

A clear methodology is essential because the factors affecting court access are complex, spatially distributed, and intertwined with socioeconomic conditions. A structured approach ensures consistent data handling, accurate measurement of access indicators, and reliable geospatial and statistical analysis. Strong methodological design also supports ethical and transparent research practices, helping produce findings that can guide policy efforts aimed at improving equitable access to civil justice.

## **Data Sources**

The majority of the data for this study will be secondary in nature, sourced from existing datasets, government records, and publicly available geospatial information. This approach allows for a comprehensive analysis of civil filing access patterns and the identification of transportation and structural barriers across different communities.

### 1. U.S. Census Bureau (ACS & Decennial Census)
Provides socioeconomic, demographic, digital access, vehicle availability, disability status, and commute-related indicators. These variables serve as core structural access measures and contextual community characteristics.

### 2. American Community Survey Microdata (IPUMS USA)
Offers individual-level records that allow more flexible construction of access-related indicators, including transportation mode, household resources, and digital connectivity.

### 3. Rural–Urban Commuting Area (RUCA) Codes – USDA
Supplies national rural–metro classifications based on commuting patterns and population density. These codes support comparative analysis across community types.

### 4. Legal Services Corporation (LSC) – Justice Gap Survey
Provides national-level survey data on perceived legal barriers, unmet civil legal needs, and self-reported obstacles to accessing the civil justice system.

### 5. National Center for State Courts (NCSC)
Offers court statistics, filing trends, and access-to-justice reports that contextualize civil filing behavior at the national level.

### 6. Eviction Lab (Princeton University)
Provides standardized civil court eviction filing data, including geographic identifiers and filing outcomes. These data help illustrate how access barriers may influence filing patterns.

### 7. FCC Broadband Data Collection
Provides national broadband availability and internet access indicators, supporting analysis of digital access as a structural barrier to civil justice.

### 8. Courthouse Locations (State Court Websites & Administrative Records)
Geocoded courthouse locations serve as anchor points for measuring transportation burden and spatial accessibility.

## **Key Variables (Anticipated)**

The following variables represent the anticipated data elements required to support a best‑case geospatial and court access analysis. All variables listed below are quantitative in nature, consisting of numeric, boolean, or categorical indicators that can be measured, compared, and modeled. Additional access‑related variables, particularly those dependent on real‑world conditions, data availability, or respondent‑reported barriers, will be identified and refined during the data collection phase.

### Court Variables
| Variable Name | Data Type | Description |
|---------------|-----------|-------------|
| case_id | string | Unique identifier for each case. |
| hearing_date | date | Scheduled court appearance date. |
| case_type | string | Civil case categories. |
| appearance_flag | boolean | Indicates whether the defendant appeared in court; may be inferred from case outcomes. |
| default_judgment_flag | boolean | Indicates whether a default judgment was issued due to non-appearance (typically civil cases). |

### Geospatial Variables
| Variable Name | Data Type | Description |
|---------------|-----------|-------------|
| geometry | Polygon / Point | Spatial representation of courthouses or census tracts; defendant geometry if address is available. |

| census_tract | string | Identifier linking individuals to demographic and socioeconomic spatial units. |
| distance_to_courthouse_km | float | Euclidean or network-based distance between residence and assigned courthouse. |
| travel_time_minutes | float | Estimated travel time using road networks or transit routing. |
| urban_rural_code | integer | Encoded classification of spatial context (e.g., RUCA codes). |

## **Feasibility, Ethical Considerations, and Reliability**

This project uses publicly available datasets and geospatial information that are already maintained by reputable national sources. Because the data is accessible, well‑documented, and compatible with common Python tools, the methodology is practical to carry out with the time, resources, and skills available. The workflow is structured, manageable, and **feasible** within the scope of the project.

**Reliability** and validity guide how indicators and results are produced. Reliability means the data and methods yield consistent results under similar conditions, while validity ensures the measures accurately represent what they are intended to capture (Andrade, 2018). Structured workflows, clear documentation, and reproducible Python notebooks help maintain accuracy and transparency. These practices reduce errors and support long‑term reproducibility, aligning with recommendations for reliable computational research (Wilson et al., 2014).

**Ethical** care is essential, especially when working with geospatial data. The project prioritizes protecting the privacy of sensitive locations and safeguarding marginalized groups by avoiding identifiable information and using aggregated indicators. Christian ethical principles, as described in GCU’s Ethical Position Statement, emphasize integrity, honesty, and respect for individuals. These values guide the project’s commitment to truthful reporting, responsible data handling, and maintaining the dignity of all communities represented in the data.

## **Data Management**

All datasets in this study come from public sources, but they will still be managed carefully to maintain integrity, transparency, and reproducibility. Each dataset will be documented and stored in a public GitHub repository, such as a project folder titled Access-Barriers-Analysis, with clear metadata, version control, and appropriate licensing. Organizing the repository in a consistent and predictable way ensures that data, scripts, and outputs remain easy to locate, understand, and reuse.

To support a reproducible workflow, the repository will follow a structured folder layout:

| Folder Name        | Purpose                                                                                 |
|--------------------|-----------------------------------------------------------------------------------------|
| data_raw/          | Original datasets exactly as downloaded; preserved for provenance.                      |
| data_processed/    | Cleaned, geocoded, merged, and analysis‑ready datasets.                                 |
| notebooks/         | Jupyter notebooks used for exploration, analysis, and documentation.                    |
| scripts/           | Python scripts for repeatable data processing, modeling, and geospatial workflows.      |
| outputs/           | Generated tables, charts, maps, and summary files.                                      |
| docs/              | Methodology notes, project description, references, and supporting documentation.       |
| metadata/          | Data dictionaries, variable descriptions, licenses, and other metadata files.           |


Although the project relies on public datasets, any sensitive information, such as precise addresses or geospatial identifiers that could reveal individual locations, will be anonymized, masked, or aggregated before analysis. Only non‑identifiable, ethically appropriate data will be stored in the public repository. This approach aligns with responsible data stewardship and supports ethical research practices throughout the project.
