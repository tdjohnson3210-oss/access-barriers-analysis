# Roadblocks to Justice: Transportation Burden and Civil Filing Access in Rural and Metropolitan Communities

---

## Introduction
For many people, getting something as basic as civil justice starts long before they ever reach a courtroom, it starts with whether they can physically get there. Even though empirical evidence shows that self‑represented litigants (SRL) still have real civil legal needs and interest in filing cases, many citizens face transportation and digital access gaps that make acting on those needs difficult. SRL often must travel to courthouses to file documents or get assistance, yet transportation access looks very different depending on where someone lives. Rural residents may face long distances, limited public transportation, and few transportation options, while metropolitan residents may deal with shorter distances but more complex or costly transit systems. This project is structured as a hypothesis‑based research problem, examining whether transportation access limit civil filing activity among SRLs.

#### Problem Statement
Since courts do not track transportation gaps, the extent to which these barriers limit civil case filings for self‑represented litigants in rural and metropolitan communities remains unknown, a gap this project intends to measure and bring awareness to.

## Objectives
This project examines whether transportation access is an overlooked barrier to civil case filing for self‑represented litigants. Traditional socioeconomic indicators do not fully explain filing disparities, and transportation gaps may contribute to suppressed participation across rural and metropolitan communities. By linking transportation conditions with civil case patterns, the study explores accessibility factors that are not reflected in current court metrics.

- Quantify transportation conditions surrounding civil courthouses, including distance and travel time.
- Compare transportation environments across rural and metropolitan areas to identify structural differences.
- Assess whether transportation gaps correspond with lower filing activity among SRL.
- Build a prediction model that identifies geographic patterns in civil justice participation and highlights where transportation access may influence filing behavior.
- Generate evidence that can inform court modernization efforts and improve equitable access to civil justice.

## Scope
This project will examine transportation access across rural and metropolitan areas, either at a national scale or within selected pilot sites. The analysis will focus on civil courts, self‑represented litigants, and transportation conditions relevant to filing activity. 

This project will not focus on individual case outcomes, legal decision‑making, or other access‑to‑justice barriers outside transportation.

## Significance
Improving access to justice resources can produce measurable economic benefits. When individuals can reach legal services and participate fully in civil proceedings, they are more likely to secure financial recoveries, avoid costly crises, and stabilize housing and income. Legal Services Corporation’s Office of Data Governance and Analysis (2025) found that civil legal aid yields an average return of about 7 dollars for every 1 dollar invested, showing that access is directly tied to economic gain. Strengthening transportation access for self‑represented litigants therefore has both equity and economic implications.

## Preliminary Research Findings
Civil case filings have increased since 2020, driven mainly by contract, small‑claims, tort, and property cases rather than a broad rise across all litigants. The National Center for State Courts conducts an annual national data collection initiative, gathering caseload information from state courts. Early 2025 signals, based on submissions from 20 states and territories, confirm that civil filings are increasing, but the growth is apparent in specific case categories. Contract, small claims, tort, and property matters are identified as the primary drivers of this rise (Court Statistics Project, 2026). Yet within these categories, the increase is largely attributable to the following litigants: creditors, landlords, and debt‑collection entities, rather than individual SRL.

#### Do SRL civil litigants still have a need or interest in filing civil cases?  
Yes. Even though SRL filings have not increased at the same pace as overall civil filing trends since 2020, the underlying need remains strong. The Justice Gap data shows that 74% of low‑income households experienced at least one civil legal problem in the past year and many faced multiple issues, indicating substantial ongoing civil legal needs . These problems often involve essentials such as housing, health care, income, and personal safety, meaning the stakes remain high even when individuals do not file cases. Yet people seek legal help for only 25% of identified problems, and 92% of those problems receive little to no assistance  (Legal Services Corporation, 2022). 

Barriers like cost concerns and uncertainty about finding or affording a lawyer limit filing behavior, despite a clear need has been identified. As a result, SRL litigants continue to have significant civil legal needs. They continue to face obstacles that prevent them from filing at the rate reflected in the mentioned post‑2020 civil filing trends (Legal Services Corporation, 2022).

## Potential Challenges
#### Privacy Barriers
A key challenge is staying transparent without exposing sensitive, personal information. As Srinareddy et al. (2024) explains, sharing too much detail or context can confuse users or unintentionally reveal private data. Even with masking, indirect identifiers can still leak information, so techniques like redaction, tokenization, and role‑based access are essential. The project should later clearly explain protections while avoiding disclosures that compromise privacy.

#### Data‑Availability Barriers 
Empirical legal research provides very limited access to underlying datasets, only about 15–17% of studies make their data available, far below the norms in other empirical fields. This lack of accessible, standardized data makes it difficult to compare fairness outcomes or validate methodological choices. Weak or inconsistent data‑sharing practices further restrict access, meaning this project may need to narrow its scope or rely on non‑standardized external data sources, which has potential to complicate the fairness evaluation and reproducibility (Matthews & Rantanen, 2025).
