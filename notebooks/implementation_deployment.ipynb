## Implementation

After developing, evaluating, and refining the access barriers models, the next phase is to implement and deploy them within a practical application. This process involves loading the finalized trained models, generating predicted outcomes, and integrating model outputs into a user-facing environment that supports the exploration and interpretation of access-to-justice barriers. For this project, deployment is performed using Streamlit, an open-source Python framework that enables data science and machine learning applications to be delivered through an interactive web-based dashboard. Streamlit Community Cloud serves as the deployment platform, allowing the application to be hosted and accessed through a web browser without requiring users to install software locally.

Supporting resources, including trained model documents, processed datasets, and application code are maintained within a public GitHub repository. GitHub provides centralized version control and serves as the source location for the files required by the application. Together, Streamlit and GitHub provide a lightweight deployment architecture that supports reproducibility, maintainability, and future expansion of the application.

The resulting dashboard allows users to explore county-level access barriers across Washington State through interactive visualizations, transportation risk classifications, projected civil filing estimates, county context indicators, and comparative statewide analyses. Rather than presenting model outputs in isolation, the application provides a practical decision-support tool that enables researchers and practitioners to examine potential access-to-justice challenges and identify communities that may experience elevated barriers to court access.

### Performance evaluation

To evaluate the efficiency of the deployed application, several performance metrics were examined, including processor utilization, memory consumption, storage requirements, and dataset load times. These measurements provide insight into the resources required to support the dashboard and help assess whether the application can operate effectively within the constraints of Streamlit Community Cloud.

Performance testing indicated that the dashboard dataset occupied approximately 0.05 MB of memory and loaded in 0.052 seconds. These results demonstrate minimal storage and processing requirements.

````python
import pandas as pd

url = (
    "https://raw.githubusercontent.com/"
    "tdjohnson3210-oss/access-barriers-analysis/main/"
    "data_processed/dashboard_data.parquet"
)

dashboard_df = pd.read_parquet(url)

print(dashboard_df.shape)
dashboard_df.head()
````

````python
import pandas as pd
import psutil
import time

start_time = time.time()

url = (
    "https://raw.githubusercontent.com/"
    "tdjohnson3210-oss/access-barriers-analysis/main/"
    "data_processed/dashboard_data.parquet"
)

dashboard_df = pd.read_parquet(url)

load_time = time.time() - start_time

memory_mb = (
    dashboard_df.memory_usage(deep=True)
    .sum() / 1024**2
)

print("PERFORMANCE SUMMARY")
print("-" * 40)
print(f"Rows: {len(dashboard_df):,}")
print(f"Columns: {len(dashboard_df.columns)}")
print(f"Dataset Size: {memory_mb:.2f} MB")
print(f"Load Time: {load_time:.3f} seconds")
print(f"CPU Usage: {psutil.cpu_percent(interval=1)}%")

ram = psutil.virtual_memory()

print(f"RAM Usage: {ram.percent}%")
print(f"Available RAM: {ram.available / (1024**3):.2f} GB")
````

### Data Pipeline

The dashboard was designed around a simple deployment architecture that minimizes infrastructure complexity while supporting future expansion. Application resources, including trained machine learning models, processed datasets, model outputs, and dashboard code, are maintained within a public GitHub repository. When the application launches, Streamlit Community Cloud retrieves the dashboard code and accesses the required data resources from the repository. The dashboard then loads the data into memory, applies user-selected filters, and generates interactive visualizations and analytical summaries for the selected county and year.

```text
ACS Data + Court Data
          ↓
   Data Processing
          ↓
  Trained Models
          ↓
dashboard_data.parquet
          ↓
 GitHub Repository
          ↓
Streamlit Community Cloud
          ↓
Interactive Dashboard
```

The Streamlit Community Cloud platform provides basic security and access management features to control who can view and interact with the deployed dashboard. ## Scalability and Maintainability

The scalability of the Access Barriers Analysis Dashboard is primarily influenced by the volume of data and analytical outputs that must be processed and displayed. The current implementation operates on a relatively small county-level dataset and demonstrates minimal storage and computational requirements. Performance testing indicated that the dashboard dataset occupied approximately 0.05 MB of memory and loaded in approximately 0.052 seconds, suggesting that the application places minimal demands on available computing resources. According to Microsoft (n.d.), performance efficiency should be evaluated through resource utilization, capacity, and the ability of a system to adapt as workloads increase. As additional states, years of data, or predictive models are incorporated, periodic monitoring of load times, memory consumption, and overall responsiveness will help ensure acceptable performance and identify when additional computing resources may be required.

The maintainability of the dashboard is supported by the separation of processed datasets, model artifacts, model outputs, and application code. This modular structure allows individual components to be updated independently and supports future expansion with minimal changes to the overall application framework. New state-level datasets can be integrated through updates to the existing data pipeline, while version control through GitHub supports reproducibility, simplifies maintenance activities, and facilitates future enhancements.

The current implementation relies on pre-trained models and precomputed outputs, and model performance may change over time as barriers evolve. To maintain predictive relevance, the models should be retrained whenever updated ACS and civil filing data become available (annually). Because the primary data sources are updated annually, an annual retraining cycle is recommended. During each update cycle, barrier scores would be recalculated, the dashboard dataset refreshed, and model performance reevaluated prior to deployment.

## Validation

Functional validation was performed by comparing dashboard outputs against the source dashboard dataset and previously generated model results, manually verifying consistency across multiple counties. Transportation risk classifications, projected civil filing estimates, and county-level metrics displayed within the application were reviewed across multiple counties and found to match expected outputs. Dashboard development code also addressed some of these data quality checks to ensure accurate representation of the underlying datasets.

Performance validation focused on latency, throughput, and resource utilization. Testing indicated that the dashboard dataset occupied approximately 0.05 MB of memory and loaded in 0.052 seconds. The application successfully processed all 156 county-level observations, resulting in an effective throughput of approximately 3,000 records per second. No noticeable lagging in responsiveness was observed during testing. These results suggest that the dashboard has minimal storage and computational requirements and can operate efficiently.

````python
import pandas as pd

url = (
    "https://raw.githubusercontent.com/"
    "tdjohnson3210-oss/access-barriers-analysis/main/"
    "data_processed/dashboard_data.parquet"
)

dashboard_df = pd.read_parquet(url)

print("FUNCTIONAL VALIDATION")
print("-" * 40)

print(
    f"Rows loaded: {len(dashboard_df)}"
)

print(
    f"Unique counties: {dashboard_df['county'].nunique()}"
)

print(
    f"Unique years: {dashboard_df['year'].nunique()}"
)

required_columns = [
    "county",
    "year",
    "actual_civil_filings",
    "predicted_civil_filings",
    "transportation_risk"
]

for col in required_columns:
    print(
        f"{col}:",
        col in dashboard_df.columns
    )

print(
    "\nMissing Values:",
    dashboard_df.isna().sum().sum()
)
````
